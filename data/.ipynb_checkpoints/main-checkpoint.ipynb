```python
import psycopg
from psycopg import sql
```

```python


with psycopg.connect("dbname=bingo user=tokeiya3 password=ingram host=blue port=5432") as conn:
    with conn.cursor() as cur:
        for i in [x*5 for x in range(1,21)]:
            cur.execute(
                sql.SQL("DROP MATERIALIZED VIEW if EXISTS {}").format(sql.Identifier(f"mv_cards_{i}"))
            )

            cur.execute(
                sql.SQL("""
                    CREATE MATERIALIZED VIEW {} AS
                    SELECT * FROM mv_nonzero WHERE num_of_cards = {}
                """).format(
                    sql.Identifier(f"mv_cards_{i}"),
                    sql.Literal(i)
                )
            )

            cur.execute(sql.SQL("REFRESH MATERIALIZED VIEW {}").format(sql.Identifier(f"mv_cards_{i}")))

            cur.execute(sql.SQL("create index "))

            print(f"Created {i} cards")


```
